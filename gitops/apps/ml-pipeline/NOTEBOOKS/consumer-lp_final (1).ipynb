```python
import os
import json
import boto3
from datetime import datetime
from confluent_kafka import Consumer, KafkaError

KAFKA_BOOTSTRAP_SERVERS = os.getenv("KAFKA_BOOTSTRAP_SERVERS", "kafka-kafka-bootstrap.kafka.svc.cluster.local:9092")
KAFKA_TOPIC = os.getenv("KAFKA_TOPIC", "wearables-lp")

conf = {
    'bootstrap.servers': KAFKA_BOOTSTRAP_SERVERS,
    'group.id': 'python-bronze-ingestor',
    'auto.offset.reset': 'earliest',
    'enable.auto.commit': True
}

consumer = Consumer(conf)
consumer.subscribe([KAFKA_TOPIC])

try:
    while True:
        # Poll for 1.0 second
        print("Polled")
        msg = consumer.poll(1.0)

        if msg is None:
            continue
        if msg.error():
            if msg.error().code() == KafkaError._PARTITION_EOF:
                continue
            else:
                print(f"Kafka Error: {msg.error()}")
                break

        # --- DATA PROCESSING (Transformation) ---
        # Mimicking your bronze_df logic
        event_data = {
            "key": msg.key().decode('utf-8') if msg.key() else None,
            "payload": msg.value().decode('utf-8'),
            "topic": msg.topic(),
            "partition": msg.partition(),
            "offset": msg.offset(),
            "kafka_timestamp": msg.timestamp()[1], # (timestamp_type, timestamp_value)
            "ingestion_timestamp": datetime.utcnow().isoformat()
        }

        print(event_data)

except KeyboardInterrupt:
    print("Stopping consumer...")
finally:
    consumer.close()
```
