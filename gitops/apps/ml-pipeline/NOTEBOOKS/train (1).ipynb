```python
import os
import shutil
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
import mlflow
import mlflow.sklearn
from mlflow.models import infer_signature

# --- 1. SETUP DATA ---
data = pd.DataFrame({"value": np.random.uniform(0, 150, 1000)})
data['health_rate'] = (data['value'] // 10) * 10

X_train = data[['value']]
y_train = data['health_rate']

# --- 2. TRAIN ---
rf = RandomForestClassifier(n_estimators=50, max_depth=3)
rf.fit(X_train, y_train)

# --- 3. SAVE LOCALLY & REGISTER ---
model_path = "./heart_rate_model_v2"
model_name = "heart_rate_model_v2"

# Clean up local folder if it exists to avoid 'FileExistsError'
if os.path.exists(model_path):
    shutil.rmtree(model_path)

# Save locally
mlflow.sklearn.save_model(sk_model=rf, path=model_path)

# Log/Register in MLflow (Optional, since you're using local loading)
with mlflow.start_run() as run:
    signature = infer_signature(X_train, rf.predict(X_train))
    mlflow.sklearn.log_model(
        sk_model=rf,
        artifact_path="model",
        signature=signature,
        registered_model_name=model_name
    )

print(f"✅ Model saved locally to: {model_path}")
print(f"✅ Model registered in MLflow as: {model_name}")
```
